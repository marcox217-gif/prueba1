2. You have previously calculated the factorial of a number - this is somewhat similar. Ask for an integer and calculate the sum from 0 up/down to the integer.
Example: If you input 5 then calculate 5 + 4 + 3 + 2 + 1 (+ 0) = 15. If input is -4 then calculate -4 + -3 + -2 + -1 (+ 0) = -10.

```python
random_int = int(input("Give me a random integer number: "))
calculus_of_the_sum = 0
    
if random_int > 0:
    while random_int >= 0:
        calculus_of_the_sum = calculus_of_the_sum + random_int
        random_int = random_int-1

elif random_int < 0:
    while random_int <= 0:
        calculus_of_the_sum = calculus_of_the_sum + random_int
        random_int = random_int+1

print(calculus_of_the_sum)
```

5. In a loop ask for numbers using the input function. Add the numbers together as the are entered. End the loop when you write STOP instead of a number. Display the sum of the numbers.
Hint: Check for STOP before you try to typecast (convert) the number.

```python
final_sum = 0
random_question = input("Tell me a number, if you want to stop type STOP instead of a number: ")

while random_question != "STOP":
    final_sum += float(random_question)
    random_question = input("Tell me a number, if you want to stop type STOP instead of a number: ")
    
print(final_sum)
```

6. Continue to ask for numbers until you don't enter anything. Then display all the numbers and their sum.
Example: The numbers 1, 2 & 3 is entered and "1 + 2 + 3 = 6" is displayed.
Hint: You need use the numbers both as string type and number type.

```python
final_string = ""
final_sum = 0

question_number = input("Please, could you give me a number: ")

while question_number != "":
    if final_string == "":
        final_string = question_number
    else:
        final_string = final_string + (" + ") + question_number
    final_sum += float(question_number)
    question_number = input("Please, could you give me a number: ")
    

print(final_string, "=", final_sum)

```

7. Ask for 2 integers. Now print the first 20 numbers of the Fibonacci sequence given by those numbers.
If the input is 1 & 3, the output would be 1 3 4 7 11 18 29 47 etc.. The next number is the sum of the 2 previous.

```python
random_int1 = int(input("Give me a random integer number: "))
random_int2 = int(input("Give me another random integer number: "))

cont = 0
fibonacci_seq = str(random_int1) + " " + str(random_int2)

while cont <=17:
    if fibonacci_seq == str(random_int1) + " " + str(random_int2):
        num_1 = random_int1
        num_2 = random_int2
        num_3 = random_int1 + random_int2
        fibonacci_seq += " " + str(num_3)
    
    else:
        num_1 = num_2
        num_2 = num_3
        num_3 = num_1 + num_2
        fibonacci_seq += " " + str(num_3)
   
    cont += 1

print(fibonacci_seq)
```
